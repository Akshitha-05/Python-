## *Type Casting*
--> Converting one datatype to another datatype is called Typecasting.

#### *int-float,complex,string,bool*
#float,complex,string,bool

```python
a=100
type(a)
```

```python
float(a),complex(a),str(a),bool(a)
```

#### *float-int,str,complex,bool*

```python
b=12.50
type(b)
```

```python
int(b),str(b),complex(b),bool(b)
```

#### *String*
- can't be converted into int.
- can't be converted into float.
- can't be converted into complex.
- can be converted into bool.

```python
name="Akshitha"
type(name)
```

```python
#int(name)
#float(name)
#complex(name)
bool(name)
```

#### *Complex ->int,float,bool,string*

```python
# # int(comp) --- not possible->error
# # float(comp) --- not possible->error
# complex -str(possible)
# complex - bool(possible)
```
