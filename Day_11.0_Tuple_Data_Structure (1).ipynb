## Tuple_Data_Structure
- Tuple is exactly same as list except that is immutable.
- once we create we can't modify.


1. Insertion order will be preserved. 
2. Duplicates are allowed.
3. Hetrogeneous object are allowed.
4. We use () to store element , sep by " ,".


```python
t=10,20,30
t
```

```python
type(t)
```

```python
t=(1,2,3,4)
type(t)
```

```python
#EMPTY TUPLE
t=()
t
```

```python
#tuple of single variable
t=(10)
type(t)
```

```python
t=()
t=12,4,65,7,8
t=10
t=10,
t=(10,)
t=(10,20,30)
```

```python
t=(10,)
type(t)
```

### *Few functions in tuple*

```python
# len ()-->the length of the tuple
t=(10,20,30,40,50,60,70,80)
```

```python
#count() --> the count of the value in tuple
t=(10,20,30,20,20,30,10,10,40,50,60,70,80)
t.count(10)
```
