## *Bytes Array_Data Type*
- It is same as bytes data type but is mutable.

```python
y=[23,60,80,34,68]
ba=bytearray(y)
```

```python
ba
```

```python
for i in ba:
    print(i)
```

```python
ba[0]
#at place of index 0
```

```python
ba[0]=100 
#we are changing the  integer at index 0 (mutable)
```

```python
for i in ba:
    print(i)
```

```python
ba[1]
# the value in (address of) index 1
```

## *Set Data Type*
- When we want to represent a group of values without duplicate.
- where order is not important then we should go for set datatype.
  1. Insertion order is not preserved.
  2. Duplicates or not allowed.
  3. Hetrogenous objects are allowed.
  4. Index concept is not applicable in Set datatype.
  5. It is Mutable collection and growable in nature.
  6. we enclosed in {}.

```python
s={10,23.5,5+9j,"pot",True,10}

type(s)
```

```python
s
```

```python
#Index concept is not applicable in Set datatype.
#s[0]
```

```python
s.add(50)
s
```

```python
s.remove(10)
s
```

##  *Dictionary Data Type*
- If we want to represent a group of values as key pair form i.e.dict
- value should be in " " or '  '.

```python
d={101:"akshitha",102:"Shravya",103:"Navya"}
d
```

```python
type(d)
```

```python
food_item={"Starter":"Paneer","Manchuriya":"Noodles"}
food_item
```

- key:value
- key is number then it will not be in "/"".
- but if it is string then key should be in ' ' /" ".
- value will be in ' ' / " ".
- Key can not be same but value can be.
- It is growable in nature(we can add and remove)

```python
food_item["full meal"]="Paneer butter Masala"
food_item
```
