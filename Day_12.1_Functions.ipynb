```python
# write the code for finding factorial
```

```python
def fact(n):
    result=1
    for i in range(1,n+1):
        result=result*i
    return result
```

```python
fact(5)
```

```python
fact(4)
```

```python
# recursive function
```

```python
def fact(n):
    if n==0:
        res=1
    else:
        res=n*fact(n-1)
    return res
    
```

```python
fact(6)
```

### *Filter() function*
- we can use filter() to filter value from the given sequence based on some condition

```python
# To find the even numbers from the list by filter()
```

```python
def isEven(x):
    if x%2==0:
        return True
    else:
        return False
l=[0,5,6,9,8,60,20,11,42,13,12]
l1=list(filter(isEven,l))
l1
```

```python
# with lambda function:
l=[0,6,9,99,56,34,87]
l1=list(filter(lambda x:x%2==0,l))
l1
```

```python
# with lambda function:
l=[0,6,9,99,56,34,87]
l1=list(filter(lambda x:x%2!=0,l))
l1
```

### *map() function*
- For every elements present in the given sequence ,apply some functionality and generate new element with the required modification for this map().

- for every element present in the list perform double and generate new list of double

```python
l=[1,2,3,4,5,6,7]
def doubleit(x):
    return 2*x
l1=list(map(doubleit,l))
l1
```

### *reduce() function*
- it reduces sequence of elements into a isngle elemnt by applying the specified function.


- reduce() function represent in functools module and hence we should write import statemwnt.

```python
from functools import *
l = [10,20,30,40,50]
result = reduce(lambda x,y:x+y,l)
print(result)
```
