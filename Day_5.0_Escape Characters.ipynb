## *In String literals we can escape character to associate a special meaning.*

```python
# \n-->for new line
da="Data\nAnalytics"
print(da)
```

```python
# \t -->for one tab distance
gd="Graphics\tdesign"
print(gd)
```

```python
# \b for bACKSPACE
JD="Mani\bpopulation"
print(JD)
```

```python
# \v for vertical tab
JD="Mani\vpopulation"
print(JD)
```

```python
a="mani\rdev"
print(a)
```

```python
print("Ice\rcream")
```

```python
b="Ice\fcream"
print(b)
```

```python
print("Ice\fcream")
```

```python
c="Ice\'cream"
print(c)
```

```python
d="Ice\\cream"
print(d)
```
