### Iterative Statements
- If we want to execute a group of statement multiple times then we should go for iterative statements.
   1. for loop
   2.  while loop    

#### *1.For Loop*
- if we want to execute some action for every element present in some sequence (collection of string) then we should go for For loop.
 

```python
#printing each element in the string
s="Virat Kohli"
for x in s:
    print(x)
```

```python
#printing each element in the list
list1=[23,"Name",6,77,68,78]
for x in list1:
    print(x)
```

#### *While Loop*
- if we want to execute a group of statements iteratively until some condtion false,then go for while loop.

```python
# to print numbers from 1 to 10 using while loop.
x=1
while x<=10:
    print(x)
    x=x+1
```

```python
# display the sum of first n numbers
n=int(input("enter the num:"))
sum=0
i=1
while i<=n:
    sum=sum+i
    i=i+1
print("the sum of first",n,"number is ",sum)
```

#### *Nested Loops*
- Loops within the loop

```python
for i in range(4):
    for j in range(4):
        print("i=",i,"j=",j)
```

### *Transfer Satements*
#### *break*
- we can use break statement inside loops to break loop
   execution based on some condition.

```python
for i in range(10):
    if i==7:
        print("Processing is enough, please break")
        break
    print(i)
```

```python
cart=[100,200,400,600,700,900]
for item in cart:
    if item>500:
        print("you will 20% discount on this order")
        break
    print(item)
    
```

#### *Continue*
- we can use continue statement to skip current iteration and continue to next.

```python
for i in range(10):
    if i%2==0:
        continue
    print(i)
```

```python
cart=[200,400,500,600,700,900,1000]
for item in cart:
    if item>=500:
        print("we can't process the item ",item)
        continue
    print(item)
```

#### *Pass*
- Pass is a keyword in python.
- in python if block is required which wo,t do anything then we can define that empty block with pass keyword.
- pass:
  1. it is an empty statement.
  2. it is null statement
  3. it won't do anything

```python
for i in range(100):
    if i%9==0:
        print(i)
    else:pass
```
