## Flow Control
- Flow control describe the order in which the statements will be executed.

### *Types of Flow Control*
1. Iterative Statements = for,while
2. Transfer Statements = break,continue,pass
3. Conditional Statements = if,else,elif

### *Condtional Statements*

#### if condition


```python
name="Akshitha"
if name=="Akshitha":print("Hi",name)
```

#### if-elif-else condition

```python
brand=input("What is your phone brand?")
if brand=="Apple":
    print("Apple is my phone brand")
elif brand=="Samsung":
    print("Samsung is my phone brand ")
elif brand=="Motorola":
    print("Motorola is my phone brand")
else:
    print("no phone")
```
