#### *How to Take multiple input at a time?*

```python
a,b=[int(x)for x in input("enter 2 numbers:").split()]
print("product of this:",a*b)
```

```python
a,b,c=[float(x) for x in input("enter 3 numbers:").split(",")]
print("product of this:",a*b*c)
```

### *Eval function():*
- eval will take a string and evaluate the result

```python
x=eval("14+85+899")
x
```

```python
#input of expression 
x=eval(input("Enter expression"))
x
```

```python
type(x)
```

```python
#input of List 
l=eval(input("enter a list"))
print(type(l))
```

#### *Formated string*

```python
name='pavan'
salary=250000
friend='sunny'
print("Hello {0} your salary is {1} and your friend is {2}".format(name,salary,friend))
print("Hello {x} your salary is {y} and your friend is {z}".format(x=name,y=salary,z=friend))
```
