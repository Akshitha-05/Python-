## ***Data Types***
- Data type arerepresent the type of data present inside a variavble.
- In Python we are not required to specify the type explicitly.
- Based on value provided, the type will be assigned automatically.
- i.e. dynamically typed language

```python
#type() ->It is an in-built func in python to show the data type of a value
a=10
type(a)
```

### ***No of Data Types(Built in) that Python contain***
- *Five Fundamental data types*
    1. int ==> numbers(-1,0,1)
    2. float ==> decimal numbers(4.5,3.21)
    3. str ==> characters('abc')
    4. complex ==> real+imaginary(5+8j)
    5. bool ==> True/False(1/0)

```python
# integer datatype
a=10
print(type(a))
#float datatype
b=2.0
print(type(b))
#string datatype
c='abc'
print(type(c))
#complex datatype
d=2+6j
print(type(d))
print(d.real)
print(d.imag)
#bool datatype
e=True
print(type(e))
```

## Conversions

### - *Decimal to all*

```python
a=255
bin(a),oct(a),hex(a)
```

### - *Binary to all*

```python
b=0b11111111
b,oct(b),hex(b)
```

### - *Octal to all*

```python
c=0o2435
c,bin(c),hex(c)
```

### - *Hex to all*

```python
d=0xAB345
d,bin(d),oct(d)
```
