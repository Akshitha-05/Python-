## FileHandling
- we have two types of file.
  1. text file -->abc.txt
  2. binary file --> imagefile,videofile,audio file

### *opening a file in python*

```python
f=open('abc.txt','w')
print('filename',f.name)
print('filename',f.mode)
print('is it readable',f.readable())
print('is it readable',f.writable())
f.close()
```

### *writing text in the file*

```python
f=open("abc.txt","w")
f.write("Akshitha\n")
f.write("Munnanuri")
print("written")
f.close()
```
