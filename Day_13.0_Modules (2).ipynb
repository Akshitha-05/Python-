### *Random module*
- It is use to generate random number.
- It is having multiple functions.

```python
from random import *
for i in range(10):
    print(random())
```

### *randint()*
- it is used to generate random integer between 2 given numbers.

```python
from random import *
for i in range(10):
    print(randint(1,20))
```

### *uniform()*
- It returns random float value between 2 numbers
- it will exclude the value what we give and it will floating points between those two numbers.
  

```python
from random import *
for i in range(10):
    print(uniform(1,10))
```

### *randrange()*


```python
import random 
for i in range(12):
    print(randrange(1,11,2))
```

### *choice()*
- it will not return random number
- it will return random object from the given list/tuple.

```python
from random import *
list1=['akki','nikki','chikki','pinni','bunny']
for i in range(10):
    print(choice(list1))
```
