 ### *1.Arithmetic operator*
  - +addition
  - -substrcation
  - (*) multiplication
  - / division
  - % modulo
  - // floor division
  - ** for power/exponential 

```python
a=10
b=2
print(a+b) #addition
print(a-b) #substraction
print(a*b) #multiplication
print(a/b) #division
print(a%b) #modulo
print(a//b) #floor division -->only integer value will be the output
print(a**b) #exponential
```
